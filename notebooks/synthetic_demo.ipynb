# Public synthetic demo only

This executes independent manufactured 3D diffusion, not the private PEMFC experiment. It contains no CFD/UDF data. Tests must pass before the demo runs. No execution results are claimed in this notebook snapshot.


```python
import subprocess, sys
from pathlib import Path
root = Path('/content/pemfc-domain-pinn')
if not root.exists():
    subprocess.run(['git', 'clone', 'https://github.com/powerfulsmooth/pemfc-domain-pinn.git', str(root)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-e', str(root) + '[test]'], check=True)
subprocess.run([sys.executable, str(root/'tools/check_public_record.py')], check=True)
subprocess.run([sys.executable, '-m', 'pytest', str(root/'tests'), '-q'], check=True)
```

```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('Synthetic scope; device:', device, 'Torch:', torch.__version__)
subprocess.run([sys.executable, str(root/'examples/train_synthetic.py'), '--device', device, '--warmup', '20', '--physics', '20', '--output', '/content/public_synthetic_demo'], check=True)
```

The synthetic output is under /content/public_synthetic_demo. Runtime reset can remove it; copy it to your own storage if needed. A trusted synthetic checkpoint can be resumed with the same settings and --resume. This does not validate PEMFC accuracy or the withheld private physics.
